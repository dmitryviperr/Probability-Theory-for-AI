# Задача 1.8. 
>В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

а)
$$
\tilde{A}_{10}^{5} = 10^5 = 100\,000.
$$
б)
$$
\tilde{C}_{10}^{5} = C_{10+5-1}^{5} = C_{14}^{5} = \frac{14 \cdot 13 \cdot 12 \cdot 11 \cdot 10}{1 \cdot 2 \cdot 3 \cdot 4 \cdot 5} = 2002.
$$

```python
import scipy.special as sc

movies = 10 
prizes = 5   

# а) 
result_a = movies ** prizes

# б) 
result_b = sc.comb(14, 5)
print(f"а) {result_a}")
print(f"б) {result_b}")
```

# Задача 1.9.

> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 — по 2 раза?

$$
P_7(3; 2; 2) = \frac{7!}{3! \cdot 2! \cdot 2!} = 210.
$$

```python
from itertools import product
counter = 0
for d in product('456', repeat = 7):
    num = ''.join(d)
    if num.count('4') == 3 and num.count('5') == 2 and num.count('6') == 2:
        counter += 1
print(counter)
```

# Задача 1.10.

> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

а)
$$
P(A) = \frac{m}{n} = \frac{1}{A_6^3} = \frac{1}{6 \cdot 5 \cdot 4} = \frac{1}{120}.
$$

б)
$$
P(B) = \frac{m}{n} = \frac{1}{P_6} = \frac{1}{6!} = \frac{1}{720}.
$$

```python
from itertools import permutations
counter_a = 0
counter_b = 0
# а)
for ch in permutations('ТЕИЯРО', 3):
    counter_a += 1


for ch in permutations('ТЕИЯРО', 6):
    counter_b += 1
result_a = 1 / counter_a
result_b = 1 / counter_b

print(f"а) {result_a:.4f}")
print(f"б) {result_b:.4f}")
```

# Задача 1.11.

> Используя условие задачи 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

$$
P(B) = \frac{m}{n} = \frac {1}{P_6(3; 2; 1)} = \frac{3! \cdot 2! \cdot 1!}{6!} = \frac{1}{60}.
$$

```python
from itertools import permutations
counter = 0
for ch in product('АНС', repeat = 6):
    word = ''.join(ch)
    if word.count('А') == 3 and word.count('Н') == 2 and word.count('С') == 1:
        counter += 1
result = 1/counter
print(result)
```

# Задача 1.16.

> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

а)
$$
P(A) = \frac{m_1}{n} = \frac{2(5!)^2}{10!} = \frac{2 \cdot 120^2}{3\,628\,800} = 0{,}00794.
$$

б)
$$
P(B) = \frac{m_2}{n} = \frac{5! \cdot 2^5}{10!} = \frac{120 \cdot 32}{3\,628\,800} = 0{,}00106.
$$

```python
import scipy.special as sc

P_A = (2 * sc.factorial(5) ** 2) / (sc.factorial(10))
P_B = (sc.factorial(5) * 2 ** 5) / (sc.factorial(10))

print(f"а) {P_A:.5f}")
print(f"б) {P_B:.5f}")
```

> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

а)
$$
P(A) = \frac{m_1}{n} = \frac{C_9^2 \cdot C_8^7}{C_{36}^7} = \frac{36 \cdot 8}{8\,347\,680} = 0{,}0000345.
$$

б)
$$
P(B) = \frac{m_2}{n} = \frac{C_9^7 \cdot 4^7}{C_{36}^7} = \frac{36 \cdot 16\,384}{8\,347\,680} = 0{,}07066.
$$

в)
$$
m_3 = C_9^3 \left( 6C_4^4 C_4^2 C_4^1 + 3C_4^3 C_4^3 C_4^1 + 3C_4^3 C_4^2 C_4^2 \right) =
$$
$$
= 84(6 \cdot 24 + 3 \cdot 64 + 3 \cdot 144) = 84 \cdot 768 = 64\,512.
$$

$$
P(D) = \frac{m_3}{n} = \frac{64\,512}{8\,347\,680} = 0{,}00773.
$$

```python
import scipy.special as sc
# а)
P_A = (sc.comb(9, 2) * sc.comb(8, 7)) / (sc.comb(36, 7))

# б) 
P_B = (sc.comb(9, 7) * 4 ** 7) / (sc.comb(36, 7))

# в)
P_D = (sc.comb(9, 3) * (6 * sc.comb(4, 4) * sc.comb(4, 2) * sc.comb(4, 1) + 
                        3 * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(4, 1) 
                        + 3 * sc.comb(4, 3) * sc.comb(4, 2) * sc.comb(4, 2))) / sc.comb(36, 7)
print(f"а) {P_A:.5f}")
print(f"б) {P_B:.5f}")
print(f"в) {P_D:.5f}")
```

# Задача 1.43.

> Наудачу взятый телефонный номер состоит из 5 цифр. Какова вероятность того, что в нем все цифры: а) различные; б) одинаковые; в) нечетные? Известно, что номер телефона не начинается с цифры ноль.

а)
$$
P(A) = \frac{m}{n} = \frac{9 \cdot 9 \cdot 8 \cdot 7 \cdot 6}{9 \cdot 10^4} = \frac{27\,216}{90\,000} = 0{,}3024.
$$

б)
$$

$$
P(C) = \frac{m}{n} = \frac{5 \cdot 5^4}{9 \cdot 10^4} = \frac{3\,125}{90\,000} = 0{,}0347.
